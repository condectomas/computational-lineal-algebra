# Jordan canonical form

In this notebook we work with a linear transformation on a complex vector space $V$,
$$
f:V\to V.
$$
The Jordan canonical form is the simplest matrix representation of a linear map: if the map is diagonalizable it reduces to a diagonal matrix; otherwise it keeps the eigenvalues on the diagonal and may include ones on the superdiagonal. Here we will explore its properties and how Jordan chains arise.

```python
import numpy as np
```

Now we compute the characteristic polynomial and the minimal polynomial of $A$. The characteristic polynomial is
$$
P_A(\lambda)=\det(A-\lambda\mathbb{I})=(\lambda-\lambda_1)^{m_1}\cdots(\lambda-\lambda_r)^{m_r},
$$
and the minimal polynomial has the same linear factors with exponents
$$
m_A(\lambda)=(\lambda-\lambda_1)^{s_1}\cdots(\lambda-\lambda_r)^{s_r},\qquad s_j\le m_j.
$$

In this problem the eigenvalues are integers, so it is important to adjust the numerical results to avoid misleading values like 2.0000001 or 0.9999999. Use `np.linalg.eig` to compute the eigenvalues, round carefully, and determine their multiplicities.

```python
val, vec = np.linalg.eig(A)
val = np.real_if_close(val, tol=1e-8)
val = np.round(val).astype(int)
print(val)
print('multiplicities:', {k: int(list(val).count(k)) for k in np.unique(val)})
```

```python
I = np.identity(9)
A1 = A - I
A2 = A + 2*I
R = np.linalg.matrix_power(A1, 2) @ np.linalg.matrix_power(A2, 3)
print('The computation with (A-I)^2 (A+2I)^3 annihilates A:', np.allclose(R, 0, atol=1e-8))
print(np.round(R, 10))
```
